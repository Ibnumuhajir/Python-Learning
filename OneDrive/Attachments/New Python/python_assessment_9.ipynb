1. Online Shopping Cart
A customer buys 3 products costing ₹450, ₹800, and ₹250. Give a 10% discount if the total exceeds ₹1,000 and an additional ₹100 discount if the customer has a membership. Calculate the final bill.


```python
class cart:
    def __init__(self,cost=[],membership=True):
        self.cost=cost
        self.membership=membership
    def add(self,price):
        self.cost.extend(price)
        return self.cost
    def total(self):
        return sum(self.cost)
        
    def final_bill(self):
        t=self.total()
        if t>1000:
            t=t-t*10/100
        if self.membership:
            t=t-100
        return t
            
        
    
a=cart()
print(a.add([450,800,250]))
a.total()
a.final_bill()



        
```

2. Food Delivery Delay
A food delivery company promises delivery in 30 minutes. Classify each delivery as On Time, Slightly Delayed (31–45 minutes), or Highly Delayed (above 45 minutes). Process multiple delivery times.


```python
class delivery:
    promise_time=30
    def __init__(self,time):
        self.time=time
    
    def delay(self):
        if self.time>45:
            print("highly delayed")
        elif self.time>30:
            print("slightly delayed")
        else:
            self.time<self.promise_time
            print("food on time")

a=delivery(46)
b=delivery(35)
c=delivery(15)
a.delay()
b.delay()
c.delay()
```

5. Product Stock Alert
Given stock quantities for several products, label each as Out of Stock, Low Stock (1–10), or Available (above 10).
Also count how many products fall into each category.

```python
a={"car":7,"bike":12,"truck":4,"auto":16,"plane":0}
out_of_stock=0
low_stock=0
available=0
for p,q in a.items():
    if q==0:
        print(p,"out of stock")
        out_of_stock+=1
    elif q<10:
        print(p,"low stock")
        low_stock+=1
    elif q>10:
        print(p,"available")
        available+=1
print("count of out of stock = ",out_of_stock)
print("count of low stock = ", low_stock)
print("count available= ", available)
```

16. Missing Data Replacement
Given temperatures = [29, None, 31, 30, None, 28, 32], replace each missing value with the average of the
available values, rounded to two decimal places. Do not use external libraries.

```python
temperatures =[29, None, 31, 30, None, 28, 32]
sum=0
length=0
for i in temperatures:
    if i is not None:
        sum+=i
        length+=1
average=sum/length
new=[int(average) if i is None else i for i in temperatures]
print("temperature =",new)
```
