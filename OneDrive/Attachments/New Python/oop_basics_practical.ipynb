So far your programs have been a list of steps. Object-oriented programming lets you bundle **data** and the **actions that work on that data** together into one thing, called an object. ## Session 3 — Giving objects actions (methods)

A method is a function written inside a class. Its first parameter is always `self`, which is how it reaches the object's own data.

```python
class Book:
    def __init__(self, title, pages):
        self.title = title
        self.pages = pages

    def describe(self):
        print(self.title, "has", self.pages, "pages")

    def reading_days(self, pages_per_day):
        return self.pages / pages_per_day

b = Book("Atomic Habits", 320)
b.describe()
print(b.reading_days(40))
```

`__str__` controls what you see when you `print()` the object itself.

```python
class Book:
    def __init__(self, title, pages):
        self.title = title
        self.pages = pages

    def __str__(self):
        return self.title + " (" + str(self.pages) + " pages)"

b = Book("Atomic Habits", 320)
print(b)
```

Methods can also **change** the object's data.

```python
class Book:
    def __init__(self, title, pages):
        self.title = title
        self.pages = pages
        self.current_page = 0

    def read(self, count):
        self.current_page = self.current_page + count

b = Book("Atomic Habits", 320)
b.read(25)
b.read(40)

print(b.current_page)
```

## Session 4 — Shared data vs personal data

Data set with `self.` inside `__init__` belongs to **one object**. Data written directly inside the class belongs to the **class itself** and is shared by every object.

```python
class Employee:
    company = "Route Upskill Academy"

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

e1 = Employee("Anu", 40000)
e2 = Employee("Ravi", 45000)

print(e1.company, "|", e2.company)
print(Employee.company)
```

```python
e1.salary = 50000

print(e1.salary)
print(e2.salary)
```

A shared value is handy for keeping count of how many objects exist.

```python
class Ticket:
    count = 0

    def __init__(self, seat):
        self.seat = seat
        Ticket.count = Ticket.count + 1

t1 = Ticket("A1")
t2 = Ticket("A2")
t3 = Ticket("A3")

print(Ticket.count)
```

## Session 5 — Protecting an object's data (Encaplsulation)

Nothing stops someone writing `account.balance = -999`. To protect important data, start the attribute name with two underscores (`__`) — Python then hides it from direct access — and let people change it only through methods that check the value first.

```python
class Wallet:
    def __init__(self, balance):
        self.__balance = balance

    def get_balance(self):
        return self.__balance

    def add(self, amount):
        if amount <= 0:
            raise ValueError("Amount must be positive")
        self.__balance = self.__balance + amount

w = Wallet(500)
w.add(200)

print(w.get_balance())
```

```python
try:
    print(w.__balance)
except AttributeError:
    print("Can't reach it directly")
```

```python
try:
    w.add(-50)
except ValueError as e:
    print(e)
```

## Session 6 — Building on an existing class (inheritance)

A new class can start from an existing one and gain all its attributes and methods automatically. The existing class is the **parent**; the new one is the **child**. This saves you rewriting the same code.

```python
class Vehicle:
    def __init__(self, brand, speed):
        self.brand = brand
        self.speed = speed

    def info(self):
        print(self.brand, "runs at", self.speed, "km/h")

class Bike(Vehicle):
    pass

b = Bike("Yamaha", 120)
b.info()
```

A child can add its own data (using `super()` to reuse the parent's setup) and can **override** a method with its own version.

```python
class Truck(Vehicle):
    def __init__(self, brand, speed, load):
        super().__init__(brand, speed)
        self.load = load

    def info(self):
        print(self.brand, "carries", self.load, "kg")

t = Truck("Tata", 80, 5000)
t.info()
print(t.speed)
```

```python
print(isinstance(t, Truck))
print(isinstance(t, Vehicle))
print(isinstance(t, Bike))
```

Because both children have an `info()` method, you can call it on any of them without caring which kind it is. This is called **polymorphism**.

```python
fleet = [Bike("Hero", 90), Truck("Eicher", 70, 3000)]

for vehicle in fleet:
    vehicle.info()
```
