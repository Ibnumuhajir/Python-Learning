```python
#7. Count how many uppercase letters, lowercase letters, and digits are present in a given string.
text="python 3.0 is awesome!"
upper=sum(1 for i in text if i.isupper())
lower=sum(1 for i in text if i.islower())
digit=sum(1 for i in text if i.isdigit())
print(f"uppercase={upper} \nlower case={lower}\ndigit={digit}")
```

```python
#10. Given a list of numbers, find the second largest number without using sorted() or max().
def second_large(a):
    first=second=float("-inf")
    for i in a:
        if i>first:
            second=first
            first=i
        elif i>second and i!=first:
            second=i
            return first
a=[12,45,25,74,69,22]
second_large(a)


```

```python
#11. Simulate rolling a die 100 times using random numbers, count how many times each face (1–6) appears, and print the frequency of each.(1–6) appears, and print the frequency of each.

import random
one=0
two=0
three=0
four=0
five=0
six=0
for i in range(100):
    dice=random.randint(1,6)
    if dice==1:
        one+=1
    elif dice==2:
        two+=1
    elif dice==3:
        three+=1
    elif dice==4:
        four+=1
    elif dice==5:
        five+=1
    else:
        six+=1
print(one)
print(two)
print(three)
print(four)
print(five)
print(six)
```
