```python
#Q10. Write a function replace_word(filename, old, new) that saves the changed text to a new file and returns how many replacements it made.
def replace(filename,old,new):
    with open(filename,"r") as file:
        a=file.read()
        count=a.count(old)
        new_content=a.replace(old,new)
    with open("new.txt","w") as file:
        file.write(new_content)
    return "total replacement",count
replace("w.txt","in","super")

```

```python
#Q11. A file has one word per line. Write the palindromes (e.g. madam, level) to palindromes.txt and print how many there were.
with open("w.txt","r") as file:
    a=file.read()
    b=a.split()
    count=0
with open("palindrom.txt","w+") as file:
    for i in b:
        if i == i[::-1]:
            count+=1
            file.write(f"{i} \n")
print("total count =",count)
```

```python
def check_balance():
    with open("transactions.txt","r") as file:
        a=file.readlines()
        balance=0
        for i in a:
            x,y=i.split(",")
            if x.strip()=="deposit":
                balance+=int(y)
            elif x.strip()=="withdraw":
                balance-=int(y)
    return balance
check_balance()
```
