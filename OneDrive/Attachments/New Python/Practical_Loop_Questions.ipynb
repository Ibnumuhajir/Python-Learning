```python
#15. Take a number and check whether it is a **prime number**.
a=int(input("enter a number:"))
if a>1:
    prime=True
    for i in range(2,a):
        if a%i==0:
            prime=False
            break
    if prime:
        print("prime number")
    else:
        print("not prime number")
else:
    print("not")
        
```

```python
#22. Ask the user to enter **10 numbers** and count how many are **positive, negative, and zero**.
positive=0
negative=0
zero=0
for i in range(10):
    a=int(input(f"enter a number {i+1}:"))
    if a>0:
        positive+=1
    elif a<0:
        negative+=1
    else:
        zero+=1
print("positive number are :",positive)
print("negative number are:",negative)
print("zeros are:",zero)
```

```python
#27. Create a simple **ATM PIN checker**.
    #Give the user a maximum of **3 attempts** to enter the correct PIN. Display `"Account Locked"` after 3 incorrect attempts
attempt=0
password=1234
while attempt<3:
    a=input("enter password (only 3 attempt)")
    attempt+=1
    if a==password:
        print("access granted")
        break
print("Account Locked")

```
