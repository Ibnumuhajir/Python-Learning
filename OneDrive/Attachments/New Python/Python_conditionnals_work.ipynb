```python
customer_type = input("Enter customer type (regular/premium): ")
severity = input("Enter issue severity (low/medium/high): ")
if customer_type == "regular":
    if severity == "low":
        priority = "Normal"
    elif severity == "medium":
        priority = "High"
    elif severity == "high":
        priority = "Urgent"
    else:
        priority = None
elif customer_type == "premium":
    if severity == "low":
        priority = "High"
    elif severity == "medium":
        priority = "Urgent"
    elif severity == "high":
        priority = "Critical"
    else:
        priority = None
else:
    priority = None

if priority:
    print(f"Ticket Priority: {priority}")
else:
    print("Invalid customer type or severity entered.")
```

```python
prices = {
    "Amazon": 18500,
    "Flipkart": 17999,
    "Croma": 19200
}
budget = 18000
lowest_seller = min(prices, key=prices.get)
lowest_price = prices[lowest_seller]
print(f"Lowest price seller: {lowest_seller} {lowest_price}")
if lowest_price <= budget:
    print("The item is within your budget!")
else:
    extra_needed = lowest_price - budget
    print(f"The price is above your budget. Additional money required: {extra_needed}")
```
