3. Create a temperature class that stores a value in Celsius and can give the same temperature in
Fahrenheit and in Kelvin. Test it with 36.6 and with 100.

```python
class temperature:
    def __init__(self,celsius):
        self.celsius=celsius
    def fahrenheit(self):
        return (self.celsius*9/5)+32
    def kelvin(self):
        return self.celsius+273.15
a=temperature(36.6)
b=temperature(100)
print(a.fahrenheit())
print(b.fahrenheit())
print(a.kelvin())
print(b.kelvin())
```

9. Create an electricity meter class that keeps its units used hidden. Units can only be added
through a method that refuses negative values. The bill is Rs. 5 per unit for the first 100 units and
Rs. 8 per unit after that. Add 60 units, then 90 more, and print the bill. Then try adding -10 units and
show the message.

```python
class electricity_meter:
    def __init__(self):
        self.__unit=0

    def add_unit(self,unit):
        if unit<0:
            print("error : unit refuse negative value ")
        else:
            self.__unit+=unit
    def calculate(self):
        unit=self.__unit
        if unit<100:
            bill=unit*5
        else:
            bill=(100*5)+((unit-100)*8)
        return bill
    def show_bill(self):
        print("total unit",self.__unit)
        print("total bill",self.calculate())
    
a=electricity_meter()
a.add_unit(60)
a.add_unit(90)
a.show_bill()
a.add_unit(-10)
a.show_bill()
```

10. Create a parking lot class with a fixed capacity. Cars can enter and leave, and the lot must
refuse with a clear message if it is full when a car arrives, or empty when a car tries to leave. Use a
capacity of 2 and show three arrivals followed by three departures.

```python
class parking:
    capacity=2
    def __init__(self,car=0):
        self.car=car
    def arrive(self):
        if self.car>=self.capacity:
            print("parking was full")
        else:
            self.car+=1
            print("car arrive a lot",self.car,"capacity",self.capacity)
    def leave(self):
        if self.car<=0:
            print("parking is lot of emty")
        else:
            self.car-=1
            print("car left",self.car,"capacity",self.capacity)
a=parking()
a.arrive()
a.leave()

      


```
